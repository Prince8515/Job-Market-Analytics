```python
import sqlite3
import pandas as pd

df=pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Minor Project/job_market_cleaned.csv')
df['via'] = df['via'].str.replace('via ', '', regex=False)
conn = sqlite3.connect("job_market_cleaned.db")

```

```python
df.to_sql(
    "jobs",
    conn,
    if_exists="replace",
    index=False
)
```

```python
pd.read_sql_query("SELECT * FROM jobs LIMIT 5;", conn)
```

SQL_ANALYSIS

1.Total job postings

```python
pd.read_sql_query("Select Count(*) AS Total_Job_Postings  FROM jobs ",conn)
```

2.Top 5 Job Roles

```python
pd.read_sql_query(
'''SELECT title, COUNT(*) AS Job_Count
   FROM jobs
   GROUP BY title
   ORDER BY Job_Count DESC
   LIMIT 5; ''',conn)
```

3.Top 5 Companies

```python
pd.read_sql_query(
'''SELECT company_name, COUNT(*) AS Job_Post_Count
   FROM jobs
   GROUP BY company_name
   ORDER BY Job_Post_Count DESC
   LIMIT 5; ''',conn)
```

4.Top Job Sources

```python
pd.read_sql_query(
'''SELECT via, COUNT(*) AS Job_Post_Count
   FROM jobs
   GROUP BY via
   ORDER BY Job_Post_Count DESC
   LIMIT 5; ''',conn)
```

5.Top 5 job schedule types by number of job postings?

```python
pd.read_sql_query(
'''SELECT  schedule_type, COUNT(*) AS Job_Post_Count
   FROM jobs
   GROUP BY  schedule_type
   ORDER BY Job_Post_Count DESC
   LIMIT 5; ''',conn)
```

6.How many jobs are marked as work-from-home/remote?

```python
pd.read_sql_query(
'''SELECT COUNT(*) AS Work_From_Home_Jobs
   FROM jobs
   WHERE work_from_home=1; ''',conn)
```

7.Average Yearly Salary by Role

```python
pd.read_sql_query(
'''SELECT title,AVG(salary_yearly) AS Avg_salary
   FROM jobs
   WHERE salary_yearly IS NOT NULL
   GROUP BY title
   HAVING COUNT(*) >= 10
   ORDER BY Avg_salary DESC
   LIMIT 10''',conn)
```

8.Job Postings by Month

```python
pd.read_sql_query(
'''SELECT
    Posted_Month,
    COUNT(*) AS Job_Count
    FROM jobs
    WHERE Posted_Month IS NOT NULL
    GROUP BY  Posted_Month
    ORDER BY Posted_Month''',conn)
```

9.Demand vs Average Salary

```python
pd.read_sql_query(
    '''
  SELECT title,
       COUNT(*) AS Job_count,
       AVG(salary_yearly) AS Avg_salary
FROM jobs
GROUP BY title
ORDER BY Job_count DESC
LIMIT 10;
''',conn
)
```

10.Remote Jobs by Job Role

```python
pd.read_sql_query(
    '''
    SELECT title, COUNT(work_from_home) AS Remote_Jobs
    FROM jobs
    WHERE work_from_home = 1
    GROUP BY title
    ORDER BY Remote_Jobs DESC
    LIMIT 10;
    ''',conn
)
```
