```python
df_cleaning['raw_posted_at'] = (
    pd.to_datetime(df_cleaning['date_time'])
    - pd.to_timedelta(
        df_cleaning['posted_at'].str.replace(' ago', '', regex=False)
    )
)
```

```python
df_cleaning['Posted_Year']=df_cleaning['posted_at'].dt.year
```

```python
df_cleaning[['salary_avg','salary_min','salary_max','salary_hourly','salary_yearly','salary_standardized']].describe()
```

```python
Q1 = df_cleaning['salary_yearly'].quantile(0.25)
Q3 = df_cleaning['salary_yearly'].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower:", lower)
print("Upper:", upper)

outliers = df_cleaning[
    (df_cleaning['salary_yearly'] < lower) |
    (df_cleaning['salary_yearly'] > upper)
]

print(outliers.shape)
outliers['salary_yearly'].sort_values()
```

```python
outliers[['title',  'salary_yearly']].sort_values(
    'salary_yearly',
    ascending=False
).tail(20)
```
